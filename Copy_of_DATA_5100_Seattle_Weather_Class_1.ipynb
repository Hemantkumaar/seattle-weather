### Impute missing values

We will replace missing values with the mean across years of values on that day.


**Design an algorithm for replacing missing values with the mean across years of values on that day.**

Step 1: Parse the data
The first step is to ensure your dataset is in a suitable format, which includes parsing the dates correctly and handling any initial data quality issues.
Read the data: Load the time series data from a file or database into a data structure, such as a dataframe in a programming language like Python (using the Pandas library).
Convert dates: Ensure the date column is in a standardized datetime format.
Identify missing values: Scan the dataset to find all missing entries. Missing values are commonly represented as "NaN" (Not a Number).
Step 2: Extract day and month
To identify seasonal patterns, isolate the month and day of the year for each data point.
Extract features: Create new columns for the month and day of the year from the datetime index. This will allow for grouping the data by a specific calendar day, regardless of the year.
Handle leap years: When dealing with daily data, be aware of February 29th. The simplest approach is to treat it as a unique day of the year. Another option is to group it with February 28th if the overall pattern is consistent.
Step 3: Compute the mean for each day
Calculate the average value for each day of the year, based on the non-missing data from all available years.
Group the data: Group the dataset by the newly created month and day columns.
Calculate the mean: Within each group, compute the mean of the observed values. The na.rm=TRUE parameter in many statistical packages ensures that missing values are ignored in this calculation.
Store the daily means: Save these daily mean values in a new data structure. This can be a dictionary or a separate table that maps each month and day to its average value.
Step 4: Impute missing values
Iterate through the original dataset and replace each missing value with the corresponding pre-calculated daily mean.
Find missing entries: Locate all rows with missing values that were identified in Step 1.
Match with daily mean: For each missing value, find its corresponding month and day.
Replace the value: Look up the mean value for that specific month and day from the daily means calculated in Step 3, and use it to replace the missing entry in the original dataset.

##### Define a column that labels each day by the day of the year: 1, 2, ..., 365.

```python
df['day_of_year'] = pd.DatetimeIndex(df['date']).day_of_year
```

```python
df.head(10)
```

##### Compute the mean precipitation for each day in Seattle, averaged across years.

```python
mean_day_precipitation = df.loc[
    df['city'] == 'SEA',
    ['precipitation', 'day_of_year']
].groupby(
    'day_of_year'
).mean()
```

```python
plt.figure(figsize=(20, 5))

sns.lineplot(data=mean_day_precipitation, x='day_of_year', y='precipitation')

plt.xlabel('Day of the year', fontsize=18)
plt.ylabel('Mean precipitation (inches)', fontsize=18)

plt.tick_params(labelsize=15)

plt.show()
```

##### Get the index of each row where precipitation is missing.

```python
indices = np.where(df['precipitation'].isna() == True)[0]
```

```python
indices
```

##### Replace each missing value with the mean on that day

```python
for index in indices:
    df.loc[index, 'precipitation'] = mean_day_precipitation.loc[df.loc[index,'day_of_year']].values[0]
```
